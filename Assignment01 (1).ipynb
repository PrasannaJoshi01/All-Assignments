```python
# --- 1. List Operations ---
print("--- List Operations ---")
```

```python
# Create a list
my_list = [10, 20, 30, 40]
print("Original List:", my_list)
```

```python
# Append an item to the list
my_list.append(50)
print("After Appending 50:", my_list)
```

```python
# Remove an item from the list
my_list.remove(20)
print("After Removing 20:", my_list)
```

```python
# --- 2. Tuple Operations ---
print("\n--- Tuple Operations ---")
```

```python
# Create a tuple
my_tuple = (100, 200, 300)
print("Original Tuple:", my_tuple)
```

```python
# Tuples are immutable, so to "append", we concatenate a new tuple
my_tuple = my_tuple + (400,)
print("After Adding 400 (via concatenation):", my_tuple)
```

```python
# To "remove" from a tuple, convert it to a list, remove the item, and convert back
temp_list = list(my_tuple)
temp_list.remove(200)
my_tuple = tuple(temp_list)
print("After Removing 200 (via list conversion):", my_tuple)
```

```python
# --- 3. Dictionary Operations ---
print("\n--- Dictionary Operations ---")
```

```python
# Create a dictionary
my_dict = {'name': 'Prasanna', 'course': 'BTech CSE', 'year': 1}
print("Original Dictionary:", my_dict)
```

```python
# Add a new key-value pair (Append)
my_dict['university'] = 'MIT-WPU'
print("After Adding 'university':", my_dict)
```

```python
# Remove a key-value pair
removed_value = my_dict.pop('year')
print("After Removing 'year':", my_dict)
```
