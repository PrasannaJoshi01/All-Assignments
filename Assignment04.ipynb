```python
import numpy as np

# Matrix A and B
A = [[1, 2], 
     [3, 4]]
B = [[5, 6], 
     [7, 8]]

# 1. Addition using Lists
result_list = [[0, 0], 
               [0, 0]]

for i in range(len(A)):
    for j in range(len(A[0])):
        result_list[i][j] = A[i][j] + B[i][j]

print("Addition using Python Lists:")
for row in result_list:
    print(row)

# 2. Addition using NumPy
np_A = np.array(A)
np_B = np.array(B)

result_numpy = np_A + np_B

print("\nAddition using NumPy Array:")
print(result_numpy)
```
